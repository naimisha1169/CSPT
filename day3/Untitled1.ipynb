```python
df["Duration_Min"] = df["Duration_Min"].fillna(df["Duration_Min"].mean())
```

```python
df["Genre"] = df["Genre"].fillna(df["Genre"].mode()[0])
```

```python
df["Budget_Million_USD"] = df["Budget_Million_USD"].fillna(df["Budget_Million_USD"].mode()[0])
df["Revenue_Million_USD"] = df["Revenue_Million_USD"].fillna(df["Revenue_Million_USD"].mode()[0])
df["IMDB_Rating"] = df["IMDB_Rating"].fillna(df["IMDB_Rating"].mode()[0])
df["Votes_Thousands"] = df["Votes_Thousands"].fillna(df["Votes_Thousands"].mode()[0])
df["Critic_Score"] = df["Critic_Score"].fillna(df["Critic_Score"].mode()[0])
df["Director_Experience_Years"] = df["Director_Experience_Years"].fillna(df["Director_Experience_Years"].mode()[0])
```
